1. Abstract
My.ml is a small machine learning project that checks a network connection record and tells
whether it is normal or an attack. The model is a Random Forest classifier trained on the
NSL-KDD dataset, which has 41 features for every connection. The text columns (protocol type,
service and flag) are converted with one-hot encoding and the number columns are scaled with
StandardScaler. All steps are kept in one scikit-learn pipeline, and the trained pipeline is saved as a
joblib file.
A simple website (Home, Predict, Result, Dashboard and History pages) is made so that the model
can be used by entering values in a form. The accuracy is printed by the training code when it is
run. An accuracy value is not written in this report because the output was not available while
preparing it.
2. Introduction
Computer networks are attacked in many ways, for example denial of service, port scanning and
unauthorised access. Checking every network connection by hand is slow, so machine learning can
be used to learn the difference between normal and attack connections from old labelled data.
In this project a model is trained on the NSL-KDD data and then connected to a website. The user
enters the details of one connection and the website shows the predicted result. The pages of the
website show the name "CyberShield" in the source files, but the project is called My.ml in this
report.
3. Problem Statement
A network connection record has many values (duration, protocol, bytes, error rates and so on). It is
difficult for a person to check each record manually, and a machine learning script alone is not easy
for a normal user to run. The problem is to build a model that classifies a connection record as
normal or attack and to give it a simple web interface.
4. Aim and Objectives
Aim
To build a machine learning model that classifies network connection records as normal or attack
and to use it through a web interface.
Objectives
• To load the NSL-KDD training and test files.
• To change the attack names into two classes: normal (0) and attack (1).
• To pre-process the data using one-hot encoding and scaling.
• To train and test a Random Forest classifier and save the model.
• To make web pages for entering data and viewing the result. 3
5. Literature Review
The KDD Cup 1999 data has been used a lot for intrusion detection. Tavallaee et al. [1] found that
this data has many repeated records and made a cleaner version called NSL-KDD, which is used in
this project. The Random Forest algorithm was given by Breiman [2], and the scikit-learn library
[3] provides its implementation along with the encoder, scaler and pipeline tools used here.
6. Existing and Proposed System
Existing system
Network records are often checked manually or with fixed rules. This is slow, and the rules have to
be changed again when new types of attacks come.
Proposed system
My.ml uses a trained Random Forest model to classify a record automatically. The model is used
through a web form, so the user does not need to run any code. It works only on values typed in the
form and does not capture live network traffic.

8. Theory
8.1 NSL-KDD Dataset
Each record in NSL-KDD describes one network connection with 41 features, a label (normal or
the name of an attack) and a difficulty value. Three features are text (protocol_type, service, flag)
and the remaining 38 are numbers.
8.2 One-Hot Encoding and Scaling
A model needs numbers, so text values like tcp or http are changed into 0/1 columns using one-hot
encoding. StandardScaler brings the number columns to a similar range (mean 0 and standard
deviation 1). 4
8.3 Random Forest
Random Forest builds many decision trees and takes the majority vote of all trees as the answer [2].
In this project it uses 100 trees, random_state 42 and balanced class weights.
8.4 Pipeline
A scikit-learn Pipeline joins the pre-processing and the classifier in one object, so the same steps
are applied automatically when a new record is predicted.
8.5 Accuracy
Accuracy is the number of correct predictions divided by the total number of predictions. It is the
only metric calculated in the code.
9. Methodology
• Read KDDTrain+.txt and KDDTest+.txt and give the 43 column names.
• Make the label: 0 for "normal" and 1 for every attack.
• Remove the label and difficulty columns, leaving 41 input features.
• Apply one-hot encoding to 3 text features and scaling to 38 number features.
• Train a Random Forest (100 trees) on the training file.
• Test it on the test file and calculate accuracy.
• Save the model as cyber_model.joblib and use it in the web pages.
